```python
import cv2
import torch
import os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

image_dir = "../../Data/JPEGImages"
mask_dir = "../../Data/SegmentationObject"

images = []
masks = []

for name in os.listdir(image_dir):

    if not name.endswith(".jpg"):
        continue

    mask_name = name.replace(".jpg", ".png")
    mask_path = os.path.join(mask_dir, mask_name)

    if not os.path.exists(mask_path):
        continue


    image = cv2.imread(
        os.path.join(image_dir, name)
    )

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )


    mask = np.array(
        Image.open(mask_path)
    )

    mask = torch.tensor(mask)

    images.append(
        torch.tensor(image)
    )

    masks.append(mask)


print("Number of images:", len(images))
print("Image shape:", images[0].shape)
print("Mask shape:", masks[0].shape)
print("Instance IDs:", torch.unique(masks[0]))
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(images[0])
plt.title("Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(masks[0])
plt.title("Multi-Class Instance Mask")
plt.axis("off")

plt.show()
```
